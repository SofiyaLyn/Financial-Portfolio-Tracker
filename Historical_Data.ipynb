```python
ticker_list = []
for ticker in tickers:
    ticker_list.append(ticker)
ticker_list
```

```python
# Tiny Excursus on downloading data every 1 hour for the past 12 months
data_frames = []

for ticker in tickers:
    hist_data = yf.Ticker(ticker).history(period="1y", interval="1h")[["Close"]]
    hist_data["Ticker"] = ticker
    hist_data.reset_index(inplace=True)
    data_frames.append(hist_data)

data = pd.concat(data_frames, ignore_index=True)
data.rename(columns={"Date": "Date"}, inplace=True)
data
```

```python
data["Datetime"] = data["Datetime"].dt.strftime("%d-%m-%Y %H:%M:%S")
data.rename(columns={"Datetime": "Date"}, inplace=True)
data.head()
```

```python
# check which ticker in data is not in amount
data_tickers = data["Ticker"].unique()
amount_tickers = amount["coinmarketcap_ticker"].unique()

for ticker in data_tickers:
    if ticker not in amount_tickers:
        print(ticker)
```

```python
data["Daily_Return"] = data.groupby("Ticker")["Close"].pct_change()
data["Daily_Return"] = data["Daily_Return"].fillna(0)
data.info()
```

# Using as Index some of the best ETFs focusing on crypto

```python
indexes = ["QQQJ", "BITQ", "DAPP"]
dataframe = []

for index in indexes:
    hist_data = yf.Ticker(index).history(period="2y", interval="1d")[["Close", "Volume"]]
    hist_data["Index"] = index 
    hist_data.reset_index(inplace=True)  
    dataframe.append(hist_data)  

index_frame = pd.concat(dataframe, ignore_index=True)
index_frame.rename(columns={"Date": "Date"}, inplace=True)
index_frame["Date"] = index_frame["Date"].dt.date
```
